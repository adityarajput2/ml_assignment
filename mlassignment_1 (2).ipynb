# QUES 5 . xplain why including Student_ID as a predictive feature may be inappropriate, even though it contains numerical values. Support your explanation with a comparison of the feature sets.


ANSWER :- `Student_ID` contains numerical values, but it is an identifier used to identify students, not a meaningful measure of their academic performance. Including it as a predictive feature may cause the model to learn misleading patterns and reduce its ability to generalize to new students.

With `Student_ID`, the feature set contains four columns: `Student_ID`, `Hours_Studied`, `Attendance`, and `Sleep_Hours`, with shape `(10, 4)`.

After excluding `Student_ID`, the feature set contains three columns: `Hours_Studied`, `Attendance`, and `Sleep_Hours`, with shape `(10, 3)`.

Therefore, `Student_ID` is excluded because the other three features provide more meaningful information for predicting `Marks`.

# QUES 1 . Inspect the DataFrame and write a reusable function that reports the number of rows, columns, column names, data types, and missing values. The function must work with other DataFrames too.

```python
import pandas as pd

def inspect_dataframe(data):
    print("Number of rows:", data.shape[0])
    print("Number of columns:", data.shape[1])

    print("\nColumn names:")
    print(data.columns.tolist())

    print("\nData types:")
    print(data.dtypes)

    print("\nMissing values in each column:")
    print(data.isnull().sum())


# Function call
inspect_dataframe(df)
```
